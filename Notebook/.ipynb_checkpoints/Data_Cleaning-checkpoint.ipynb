```python
# Drop unnecessary publication date columns
columns_to_drop = [
    'Unnamed: 0',
    'publish_date',
    'first_publish_date',
    'series',
    'awards',
    'characters',
    'places',
    'isbn',
    'isbn13'
]

# Drop columns only if they exist in the dataset
existing_columns = [
    col for col in columns_to_drop
    if col in df.columns
]

df = df.drop(columns=existing_columns)

# Display removed and remaining columns
print("Removed Columns:", existing_columns)
print("Remaining Columns:", df.columns.tolist())


```

```python
# Identifu numerical columns in the dataset
numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

# Identify categorical columns in the dataset
categorical_columns = df.select_dtypes(
    include='object'
).columns.tolist()

# Display numerical and categorical columns
print("Numerical Columns:\n", numerical_columns)

print("\nCategorical Columns:\n", categorical_columns)
```

```python
# Calculate the number of missing values in each column
missing_count = df.isnull().sum()

# Calculate the total number of missing values 
total_missing_values = missing_count.sum()

#Calculate the percentage of missing values in each column
missing_percentage = (missing_count / len(df)) * 100

# create a dataframe for missing values in each column
missing_analysis = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percentage': missing_percentage
})

# Sort columns by missing percemntage in descending order
missing_analysis = missing_analysis.sort_values(
    by='Missing Percentage',
    ascending=False
)

# Display missing value analytics
print(missing_analysis)
```

```python
# Calculate missing values in numerical columns
numerical_missing_values = df[numerical_columns].isnull().sum()

#Display missing values in numerical columns
print("Numerical Missing Values : \n ", numerical_missing_values)
```

```python
# Identify numerical columns containing missing values
numerical_missing_columns = df[numerical_columns].columns[
    df[numerical_columns].isnull().any()
].tolist()

# Calculate the median of numerical columns with missing values
numerical_medians = df[numerical_missing_columns].median()


# fill missing values with the respective column median
for column in numerical_missing_columns:

    df[column] = df[column].fillna(numerical_medians[column])

```

```python
# Check missing values in numerical columns after filling
numerical_missing_check = df[numerical_columns].isnull().sum()

# Display remaining missing values
print("Numerical Missing Values After Filling:")
print(numerical_missing_check)
```

```python
# Calculate missing values in categorical columns
categorical_missing_values = df[categorical_columns].isnull().sum()

# Display missing values in categorical columns
print("Categorical Missing Values :\n  ", categorical_missing_values)
```

```python
# Identify categorical columns containing missing values 
categorical_missing_columns = df[categorical_columns].columns[
    df[categorical_columns].isnull().any()
    ].tolist()

# Define a placeholder for missing categorical values
missing_placeholder = 'Unknown'


# Fill missing values with 'Unknown' in categorical columns
for column in categorical_missing_columns:
    df[column] = df[column].fillna(missing_placeholder)
```

```python
#check missing values in categorical columns after filling
categorical_missing_check = df[categorical_columns].isnull().sum()

# Display remaining missing values
print("Categorical Missing Values After Filling:")
print(categorical_missing_check)
```

```python

# Perform outlier analysis on all numerical columns using the IQR method
outlier_summary = {}

# Calculate quartiles, IQR, lower and upper limits, and count outliers
for column in numerical_columns:

     
    Q1 = df[column].quantile(0.25)

    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - (1.5 * IQR)

    upper_limit = Q3 + (1.5 * IQR)

    outlier_condition = (
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    )

    outlier_count = outlier_condition.sum()

    outlier_summary[column] = {
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Limit': lower_limit,
        'Upper Limit': upper_limit,
        'Outlier Count': outlier_count
    }

# Store the results in a DataFrame for comparison
outlier_analysis = pd.DataFrame(outlier_summary).T

outlier_analysis
```

```python
# Check for invalid average ratings outside the allowed range (0 to 5)
rating_column = 'avg_rating'

# Identify and count records with invalid ratings
if rating_column in df.columns:

    invalid_rating_condition = (
        (df[rating_column] < 0) |
        (df[rating_column] > 5)
    )

    invalid_ratings = df[invalid_rating_condition]

    invalid_rating_count = len(invalid_ratings)

    # Display invalid rating records 
    print("Invalid Ratings:", invalid_rating_count)

    print("\nInvalid Rating Records:")
    print(invalid_ratings[['title', 'avg_rating']])
```

```python
# Perform negative value analysis on all numerical columns
# Count negative values in each column
# Store and display the negative value summary

negative_value_summary = {}

for column in numerical_columns:

    negative_value_count = (df[column] < 0).sum()

    negative_value_summary[column] = negative_value_count

negative_values_analysis = pd.Series(
    negative_value_summary,
    name='Negative Value Count'
)

negative_value_summary
```

```python
# Identify all text (object) columns in the dataset
# Display the list of text columns

text_columns = df.select_dtypes(
    include='object'
).columns.tolist()

print("Text Columns:\n", text_columns)
```

```python
# Remove leading and trailing whitespace from text columns
# Display the cleaned text columns to verify standardisation

for column in text_columns:

    df[column] = df[column].apply(
        lambda value: value.strip()
        if isinstance(value, str)
        else value
    )

# Display cleaned text columns
print("Text Standardisation Completed!")
print(df[text_columns].head(10))
```

```python
# Remove extra spaces between words in text columns
# Standardise text formatting across the dataset

for column in text_columns:

    df[column] = df[column].apply(
        lambda value: ' '.join(value.split())
        if isinstance(value, str)
        else value
    )

# Display standardised text columns
print("Text Standardisation Completed!")

print("\nCleaned Text Data:")
print(df[text_columns].head(10))
```

```python
# Verify whether any extra spaces remain in text columns
# Count values containing extra spaces in each column
# Display the remaining extra space counts

print("\nChecking Extra Spaces:")

for column in text_columns:
    remaining_spaces = df[column].apply(
        lambda value: isinstance(value, str) and value != ' '.join(value.split())
    ).sum()

    print(f"{column}: {remaining_spaces} values with extra spaces")
```

```python
# Standardise author names by converting them to title case
# Display the standardised author names
# Check whether the author column exists in the dataset


author_column = 'author'

if author_column in df.columns:

    df[author_column] = df[author_column].str.title()

    print("Author Names Standardised Successfully!")

    print("\nStandardised Author Names:")
    print(df[author_column].head(10))

else:
    print("Author column not found!")
```

```python
# Compare the original and cleaned dataset dimensions
# Calculate the number of rows and columns removed
# Display the dataset cleaning summary

original_shape = df.shape

cleaned_shape = df.shape

rows_removed = original_shape[0] - cleaned_shape[0]

columns_removed = original_shape[1] - cleaned_shape[1]

print("Original Shape:", original_shape)

print("Cleaned Shape:", cleaned_shape)

print("Rows Removed:", rows_removed)

print("Columns Removed:", columns_removed)
```
